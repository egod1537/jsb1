# JSB1 roll-hold exploration
This notebook inspects one MCAP run while keeping reusable algorithms in `jsb1_analysis`. Set `JSB1_MCAP_PATH` before starting JupyterLab when needed.

```python
import os
from pathlib import Path

from jsb1_analysis.analyzers.roll_hold import RollHoldAnalyzer, RollHoldConfig
from jsb1_analysis.io.mcap import load_mcap
from jsb1_analysis.metrics.frequency import dominant_frequency, frequency_spectrum
from jsb1_analysis.plotting.timeseries import plot_roll_hold
```

```python
mcap_path = Path(os.environ.get("JSB1_MCAP_PATH", "../sample/run.mcap"))
run = load_mcap(mcap_path)
run.available_signals()
```

```python
config = RollHoldConfig()
plot_roll_hold(run, config=config);
```

```python
start_time, end_time = 20.0, 80.0
segment = run.slice(start_time, end_time)
plot_roll_hold(segment, config=config);
```

```python
spectrum = frequency_spectrum(segment.time, segment.signal(config.roll_rate_signal))
dominant_frequency(segment.time, segment.signal(config.roll_rate_signal))
```

```python
result = RollHoldAnalyzer(config).analyze(segment)
result
```
